```python
import napari
import tifffile
import numpy as np

arr = tifffile.imread("../sample_data/cell_3710.tif")
nuc = arr[55]  # uint16 label map

viewer = napari.Viewer()
viewer.add_image(arr[0], name="PCNA")          # 参考图
viewer.add_labels(nuc, name="nuc_mask")        # label map，每个细胞不同颜色

# 标出图像正中心
H, W = nuc.shape
viewer.add_points([[H//2, W//2]], name="center", size=5, face_color="red")

napari.run()
```
